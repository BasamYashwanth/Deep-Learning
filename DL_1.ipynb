```python
import torch
import torch.nn as nn
from sklearn.datasets import make_moons
import numpy as np
# Generate 300 samples with slight noise
X, y = make_moons(
    n_samples=300,
    noise=0.15,
    random_state=42
)
import pandas as pd

# Create DataFrame
df = pd.DataFrame(X, columns=["Feature1", "Feature2"])

# Add target class
df["Class"] = y

# Display first 20 rows
print(df.head(10))
#Convert input features into float tensors
X = torch.tensor(X, dtype=torch.float32)

# Convert output labels into float tensors
# unsqueeze(1) changes shape from (300,) to (300,1)
y = torch.tensor(y, dtype=torch.float32).unsqueeze(1)
class MLP(nn.Module):
    # Constructor
    def __init__(self,
                 input_dim=2,
                 hidden1=16,
                 hidden2=8):

        super().__init__()

        self.model = nn.Sequential(

            # Input Layer → Hidden Layer 1
            nn.Linear(input_dim, hidden1),
            nn.ReLU(),

            # Hidden Layer 1 → Hidden Layer 2
            nn.Linear(hidden1, hidden2),
            nn.ReLU(),

            # Hidden Layer 2 → Output Layer
            nn.Linear(hidden2, 1),

            # Convert output into probability
            nn.Sigmoid()
        )

    # Forward Propagation
    def forward(self, x):

        return self.model(x)
net = MLP()
# Print Network Architecture
print(net)
criterion = nn.BCELoss()
optimizer = torch.optim.Adam(
    net.parameters(),
    lr=0.01
)
epochs = 200

for epoch in range(epochs):

    # -------------------------------
    # Forward Propagation
    # -------------------------------

    predictions = net(X)

    # -------------------------------
    # Compute Loss
    # -------------------------------

    loss = criterion(predictions, y)

    # -------------------------------
    # Clear Previous Gradients
    # -------------------------------

    optimizer.zero_grad()

    # -------------------------------
    # Backpropagation
    # -------------------------------

    loss.backward()

    # -------------------------------
    # Update Weights
    # -------------------------------

    optimizer.step()

    # -------------------------------
    # Calculate Accuracy
    # -------------------------------

    predicted_classes = (predictions > 0.5).float()

    accuracy = (
        (predicted_classes == y)
        .float()
        .mean()
    )

    # -------------------------------
    # Print Training Progress
    # -------------------------------

    if (epoch + 1) % 20 == 0:

        print(
            f"Epoch [{epoch+1:3d}/{epochs}] "
            f"Loss = {loss.item():.4f}   "
            f"Accuracy = {accuracy.item()*100:.2f}%"
        )

with torch.no_grad():

    predictions = net(X)

    predicted_classes = (predictions > 0.5).float()

    accuracy = (
        (predicted_classes == y)
        .float()
        .mean()
    )
print("\n===================================")
print("Training Completed Successfully")
print("===================================")
print(f"Final Accuracy : {accuracy.item()*100:.2f}%")
print("===================================")
```
