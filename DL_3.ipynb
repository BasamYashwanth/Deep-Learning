```python
import numpy as np

np.random.seed(0)
X = np.random.rand(100, 1) * 10
y_true = 3 * X + 2 + np.random.randn(100, 1) * 0.5

w = np.random.randn(1)
b = np.random.randn(1)
lr = 0.01

# Step: Stochastic Gradient Descent - update after EVERY individual sample
for epoch in range(20):
    indices = np.random.permutation(len(X))
    for i in indices:
        xi, yi = X[i], y_true[i]
        y_pred_i = w * xi + b
        error_i = y_pred_i - yi

        dw_i = 2 * error_i * xi
        db_i = 2 * error_i

        w = w - lr * dw_i
        b = b - lr * db_i

    full_pred = w * X + b
    epoch_loss = np.mean((full_pred - y_true)**2)
    print(f"Epoch {epoch+1}: w={w[0]:.4f}, b={b[0]:.4f}, full-data loss={epoch_loss:.4f}")

print(f"\nFinal learned w={w[0]:.4f} (target=3), b={b[0]:.4f} (target=2)")
```
