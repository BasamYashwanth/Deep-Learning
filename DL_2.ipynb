```python
import numpy as np
# Step 1: Generate synthetic linear data y = 3x + 2 + noise
np.random.seed(0)
X = np.random.rand(100, 1) * 10
y_true = 3 * X + 2 + np.random.randn(100, 1) * 0.5

# Step 2: Initialize parameters randomly
w = np.random.randn(1)
b = np.random.randn(1)
lr = 0.01
n = len(X)

# Step 3: Manual Batch Gradient Descent loop (no autograd, no optimizer library)
for epoch in range(200):
    y_pred = w * X + b
    error = y_pred - y_true

    # Manually derived gradients of Mean Squared Error w.r.t. w and b
    dw = (2/n) * np.sum(error * X)
    db = (2/n) * np.sum(error)

    # THE gradient descent update rule
    w = w - lr * dw
    b = b - lr * db

    if (epoch+1) % 50 == 0:
        loss = np.mean(error**2)
        print(f"Epoch {epoch+1}: w={w[0]:.4f}, b={b[0]:.4f}, loss={loss:.4f}")

print(f"\nFinal learned w={w[0]:.4f} (target=3), b={b[0]:.4f} (target=2)")
```
