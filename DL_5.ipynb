```python
import numpy as np

# ---------------------------------------------------------
# 1. Input data and target
# ---------------------------------------------------------
X = np.array([[0.5],
              [1.0]])

y = np.array([[1.0]])

# ---------------------------------------------------------
# 2. Initialize weights and biases
# ---------------------------------------------------------
# Input layer -> Hidden layer
W1 = np.array([[0.4, 0.3],
               [0.2, 0.5]])

b1 = np.array([[0.1],
               [0.1]])

# Hidden layer -> Output layer
W2 = np.array([[0.6, 0.7]])

b2 = np.array([[0.2]])

learning_rate = 0.1


# ---------------------------------------------------------
# 3. Activation functions
# ---------------------------------------------------------
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


def sigmoid_derivative(a):
    return a * (1 - a)


# ---------------------------------------------------------
# 4. FORWARD PASS
# ---------------------------------------------------------

# Hidden layer weighted sum
Z1 = np.dot(W1, X) + b1

# Hidden layer activation
A1 = sigmoid(Z1)

# Output layer weighted sum
Z2 = np.dot(W2, A1) + b2

# Output prediction
A2 = sigmoid(Z2)

# Binary Cross Entropy loss
loss = -(y * np.log(A2) + (1 - y) * np.log(1 - A2))


print("========== FORWARD PASS ==========")

print("\nInput X:")
print(X)

print("\nHidden Layer Weighted Sum (Z1):")
print(Z1)

print("\nHidden Layer Activation (A1):")
print(A1)

print("\nOutput Layer Weighted Sum (Z2):")
print(Z2)

print("\nPredicted Output (A2):")
print(A2)

print("\nTarget Output (y):")
print(y)

print("\nLoss:")
print(loss)


# ---------------------------------------------------------
# 5. BACKWARD PASS
# ---------------------------------------------------------

# Output layer error
dZ2 = A2 - y

# Gradient of output weights
dW2 = np.dot(dZ2, A1.T)

# Gradient of output bias
db2 = dZ2

# Propagate error to hidden layer
dA1 = np.dot(W2.T, dZ2)

# Apply derivative of sigmoid
dZ1 = dA1 * sigmoid_derivative(A1)

# Gradient of input-to-hidden weights
dW1 = np.dot(dZ1, X.T)

# Gradient of hidden layer bias
db1 = dZ1


print("\n========== BACKWARD PASS ==========")

print("\nOutput Layer Error (dZ2):")
print(dZ2)

print("\nGradient of W2 (dW2):")
print(dW2)

print("\nGradient of b2 (db2):")
print(db2)

print("\nHidden Layer Error (dZ1):")
print(dZ1)

print("\nGradient of W1 (dW1):")
print(dW1)

print("\nGradient of b1 (db1):")
print(db1)


# ---------------------------------------------------------
# 6. UPDATE WEIGHTS AND BIASES
# ---------------------------------------------------------

W2 = W2 - learning_rate * dW2
b2 = b2 - learning_rate * db2

W1 = W1 - learning_rate * dW1
b1 = b1 - learning_rate * db1


print("\n========== UPDATED PARAMETERS ==========")

print("\nUpdated W1:")
print(W1)

print("\nUpdated b1:")
print(b1)

print("\nUpdated W2:")
print(W2)

print("\nUpdated b2:")
print(b2)
```
